```python
#this will copy our dataset and paste in clean_df
clean_df = df.copy()

print("Working dataset created successfully")
print(clean_df.shape)
#this code is to divide patients into different age groups
clean_df['AgeGroup'] = pd.cut(
    clean_df['Age'],
    bins=[-1, 17, 35, 50, 65, 120],
    labels=['0-17', '18-35', '36-50', '51-65', 'Above 65']
)

print(clean_df['AgeGroup'].value_counts())
```

```python
# first we need to change appoinment day in datetime
clean_df['AppointmentDay'] = pd.to_datetime(
    clean_df['AppointmentDay'],
    errors='coerce',
    utc=True
)
```

```python
print(clean_df['AppointmentDay'].dtype)
```

```python
# to find the appointment on per day
clean_df['AppointmentWeekday'] = clean_df['AppointmentDay'].dt.day_name()

print(clean_df['AppointmentWeekday'].value_counts())
```

```python
clean_df[['AppointmentDay', 'AppointmentWeekday']].head(10)
```

```python
clean_df['ScheduledDay'] = pd.to_datetime(
    clean_df['ScheduledDay'], errors='coerce', utc=True
)

clean_df['AppointmentDay'] = pd.to_datetime(
    clean_df['AppointmentDay'], errors='coerce', utc=True
)

clean_df['WaitingDays'] = (
    clean_df['AppointmentDay'].dt.normalize() -
    clean_df['ScheduledDay'].dt.normalize()
).dt.days

print(clean_df['WaitingDays'].head())
```

```python
#this Classifies patients based on how long they wait for appointments.
clean_df['WaitingCategory'] = np.select(
    [
        clean_df['WaitingDays'] == 0,
        clean_df['WaitingDays'].between(1, 7),
        clean_df['WaitingDays'].between(8, 30),
        clean_df['WaitingDays'] > 30
    ],
    [
        'Same Day',
        'Short Wait',
        'Medium Wait',
        'Long Wait'
    ],
    default='Invalid'
)

print(clean_df['WaitingCategory'].value_counts())
```

```python
clean_df[['WaitingDays', 'WaitingCategory']].head(10)
```

```python
# it will appointment in months
clean_df['AppointmentMonth'] = clean_df['AppointmentDay'].dt.month_name()

print(clean_df['AppointmentMonth'].value_counts())
```

```python
#Finds the percentage of appointments that patients missed
no_show_rate = (clean_df['No-show'] == 'Yes').mean() * 100

print("Overall No-Show Rate:", round(no_show_rate, 2), "%")
```

```python
#Compares the no-show rate between male and female patients.
gender_analysis = clean_df.groupby('Gender')['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print(gender_analysis)
```

```python
#Identifies which age groups have higher appointment no-show rates.
age_analysis = clean_df.groupby('AgeGroup', observed=True)['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print(age_analysis)
```

```python
#Compares appointment attendance across weekdays.
weekday_analysis = clean_df.groupby('AppointmentWeekday')['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print(weekday_analysis)
```

```python
waiting_analysis = clean_df.groupby('WaitingCategory', observed=True)['No-show'].apply(
    lambda x: (x == 'Yes').mean() * 100
)

print(waiting_analysis)
```

```python
#Appointment attendance patterns across weekdays.
weekday_analysis.plot(kind='bar')

plt.title('No-Show Rate by Appointment Weekday')
plt.xlabel('Weekday')
plt.ylabel('No-Show Rate (%)')
plt.xticks(rotation=45)

plt.show()
```
